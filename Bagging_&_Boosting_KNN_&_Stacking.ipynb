Question 1 : What is the fundamental idea behind ensemble techniques? How does
bagging differ from boosting in terms of approach and objective?


--> As we know, Ensemble learning helps improve machine learning results by combining several models. This approach allows the production of better predictive performance compared to a single model. Basic idea is to learn a set of classifiers (experts) and to allow them to vote. Bagging and Boosting are two types of Ensemble Learning. These two decrease the variance of a single estimate as they combine several estimates from different models. So the result may be a model with higher stability. Let's understand these two terms in a glimpse.

Bagging: It is a homogeneous weak learners’ model that learns from each other independently in parallel and combines them for determining the model average.
Boosting: It is also a homogeneous weak learners’ model but works differently from Bagging. In this model, learners learn sequentially and adaptively to improve model predictions of a learning algorithm.
Let’s look at both of them in detail and understand the Difference between Bagging and Boosting.

Bagging
Bootstrap Aggregating, also known as bagging, is a machine learning ensemble meta-algorithm designed to improve the stability and accuracy of machine learning algorithms used in statistical classification and regression. It decreases the variance and helps to avoid overfitting. It is usually applied to decision tree methods. Bagging is a special case of the model averaging approach.

Description of the Technique

Suppose a set D of d tuples, at each iteration i, a training set Di of d tuples is selected via row sampling with a replacement method (i.e., there can be repetitive elements from different d tuples) from D (i.e., bootstrap). Then a classifier model Mi is learned for each training set D < i. Each classifier Mi returns its class prediction. The bagged classifier M* counts the votes and assigns the class with the most votes to X (unknown sample).

Implementation Steps of Bagging

Step 1: Multiple subsets are created from the original data set with equal tuples, selecting observations with replacement.
Step 2: A base model is created on each of these subsets.
Step 3: Each model is learned in parallel with each training set and independent of each other.
Step 4: The final predictions are determined by combining the predictions from all the models.

Boosting
Boosting is an ensemble modeling technique designed to create a strong classifier by combining multiple weak classifiers. The process involves building models sequentially, where each new model aims to correct the errors made by the previous ones.

Initially, a model is built using the training data.
Subsequent models are then trained to address the mistakes of their predecessors.
boosting assigns weights to the data points in the original dataset.
Higher weights: Instances that were misclassified by the previous model receive higher weights.
Lower weights: Instances that were correctly classified receive lower weights.
Training on weighted data: The subsequent model learns from the weighted dataset, focusing its attention on harder-to-learn examples (those with higher weights).
This iterative process continues until:
The entire training dataset is accurately predicted, or
A predefined maximum number of models is reached.
Boosting Algorithms

There are several boosting algorithms. The original ones, proposed by Robert Schapire and Yoav Freund were not adaptive and could not take full advantage of the weak learners. Schapire and Freund then developed AdaBoost, an adaptive boosting algorithm that won the prestigious Gödel Prize. AdaBoost was the first really successful boosting algorithm developed for the purpose of binary classification. AdaBoost is short for Adaptive Boosting and is a very popular boosting technique that combines multiple “weak classifiers” into a single “strong classifier”.

Question 2: Explain how the Random Forest Classifier reduces overfitting compared to
a single decision tree. Mention the role of two key hyperparameters in this process.


-->  The Random Forest Classifier reduces overfitting compared to a single decision tree through the principles of ensemble learning, specifically bagging (Bootstrap Aggregating) and random subspace selection.
A single decision tree, if allowed to grow to full depth, can become highly complex and fit the training data too closely, including noise and outliers, leading to poor generalization on unseen data (overfitting). Random Forest mitigates this by:
Ensemble of Weak Learners: Instead of a single, potentially overfit tree, Random Forest builds an ensemble of multiple decision trees, each trained on a different subset of the data and features.
Decorrelation of Trees: The randomness introduced during tree construction (bagging and random subspace) ensures that the individual trees in the forest are diverse and less correlated. This decorrelation is crucial because averaging the predictions of highly correlated trees would not significantly reduce variance.
Variance Reduction: By averaging (or taking a majority vote for classification) the predictions of numerous decorrelated trees, the Random Forest significantly reduces the overall variance of the model, leading to better generalization and reduced overfitting. The errors of individual trees tend to cancel each other out.
Two key hyperparameters playing a significant role in this process are:
n_estimators (Number of Trees): This hyperparameter determines the number of decision trees in the forest. A higher number of trees generally leads to a more robust model with reduced variance and better generalization, as the averaging effect becomes more pronounced. However, increasing n_estimators also increases computational cost.
max_features (Maximum Features for Splitting): This hyperparameter controls the number of features considered when looking for the best split at each node of a decision tree. By randomly selecting a subset of features for each split, max_features introduces further randomness and helps decorrelate the individual trees, preventing them from all relying on the same dominant features and further reducing overfitting. Common values include sqrt (square root of total features) or log2 (base-2 logarithm of total features).


Question 3: What is Stacking in ensemble learning? How does it differ from traditional
bagging/boosting methods? Provide a simple example use case.

--> Stacking, or stacked generalization, is an ensemble learning technique that combines the predictions of multiple diverse models (base models) using a meta-learner (or meta-model) to make a final prediction. The base models are trained on the original dataset, and their predictions become the input features for the meta-learner, which then learns how to optimally combine these predictions.
Differences from Bagging and Boosting:
Model Diversity: Stacking typically uses heterogeneous base models (different algorithms like Logistic Regression, Decision Trees, SVMs), while bagging (e.g., Random Forest) and boosting (e.g., AdaBoost, XGBoost) usually employ homogeneous base models (multiple instances of the same algorithm, often decision trees).
Combination Method: Bagging combines predictions through averaging (for regression) or majority voting (for classification). Boosting sequentially builds models, with each new model attempting to correct the errors of the previous ones. Stacking, in contrast, trains a separate meta-learner to learn the optimal way to combine the predictions of the base models.
Objective: Bagging primarily aims to reduce variance, boosting aims to reduce bias, and stacking aims to improve overall prediction accuracy by leveraging the strengths of multiple diverse models.
Simple Example Use Case:
Consider a classification problem, such as predicting whether an email is spam or not.
Base Models: Train three different base models on the email dataset:
A Logistic Regression model.
A Decision Tree classifier.
A Support Vector Machine (SVM).
Meta-learner Input: For each email in the training set, obtain the predictions (or probabilities) from each of these three base models. These predictions (e.g., probability of being spam from Logistic Regression, Decision Tree, and SVM) now form a new set of input features.
Meta-learner Training: Train a meta-learner (e.g., another Logistic Regression or a simple Neural Network) on these new input features, with the original "spam/not spam" labels as the target. The meta-learner learns how to weight or combine the predictions of the base models to make the final classification.
Prediction: When a new email arrives, pass it through the trained base models to get their individual predictions. Then, feed these predictions into the trained meta-learner to obtain the final "spam/not spam" classification.


Question 4:What is the OOB Score in Random Forest, and why is it useful? How does
it help in model evaluation without a separate validation set?

--> he Out-of-Bag (OOB) score in a Random Forest is a performance metric calculated using data points that were not used to train each individual tree in the forest. It is useful because it provides an unbiased estimate of the model's generalization error without needing a separate validation set, making it ideal for scenarios where data is limited.  
How OOB works
Bootstrap sampling: When a Random Forest is built, each tree is trained on a bootstrap sample (a random sample with replacement) of the original training data.
Out-of-bag samples: On average, about one-third of the data points are left out of the bootstrap sample for each tree. These are called "out-of-bag" samples for that specific tree.
Prediction and aggregation: For each data point, a prediction is made using only the trees that did not use that point in their training set. These individual predictions are then aggregated (e.g., through a majority vote for classification) to get a final prediction for the OOB sample.
Score calculation: The final predictions are compared to the actual outcomes of the out-of-bag samples. This comparison is used to calculate the OOB score, often an accuracy or error rate, which represents how well the model performs on unseen data.
How it helps evaluate a model without a separate validation set
In-built validation: The OOB evaluation treats the out-of-bag samples as if they were a validation set, allowing you to evaluate the model's performance on data it has not seen during training.
Eliminates the need for a validation set: This internal validation process means you do not need to split your data into separate training and validation sets, which is especially helpful when you have a limited amount of data.
Provides an unbiased estimate: By using data that was intentionally left out of the training for each tree, the OOB score provides a reliable and unbiased estimate of how the model will perform on new, unseen data.
Acts as a sanity check: It can be used during the training process to ensure the model is still learning. If the OOB error stops decreasing, it can indicate that adding more trees is no longer improving the model.


Question 5: Compare AdaBoost and Gradient Boosting in terms of:
● How they handle errors from weak learners
● Weight adjustment mechanism
● Typical use cases

--> AdaBoost and Gradient Boosting are both ensemble methods that combine multiple weak learners to create a strong learner, but they differ in their approach to handling errors and adjusting weights.
1. How they handle errors from weak learners:
AdaBoost: AdaBoost focuses on misclassified samples. In each iteration, it identifies the samples that were misclassified by the previous weak learner and increases their weights, ensuring that subsequent weak learners pay more attention to these difficult instances. The algorithm specifically aims to reduce classification errors.
Gradient Boosting: Gradient Boosting focuses on minimizing a loss function by training subsequent weak learners on the residuals (the difference between the actual and predicted values) of the previous models. It effectively tries to correct the errors of the preceding models by fitting new models to the negative gradient of the loss function. This allows it to handle both classification and regression tasks.
2. Weight adjustment mechanism:
AdaBoost: AdaBoost adjusts the weights of the training samples based on their classification accuracy. Misclassified samples receive higher weights, while correctly classified samples receive lower weights. Additionally, each weak learner is assigned a weight in the final ensemble based on its accuracy.
Gradient Boosting: Gradient Boosting does not directly adjust sample weights in the same way as AdaBoost. Instead, it assigns a weight (or learning rate) to the predictions of each weak learner when adding them to the ensemble. This weight is typically a small value that controls the contribution of each new learner and helps prevent overfitting. The focus is on fitting the residuals, not reweighting samples.
3. Typical use cases:
AdaBoost: AdaBoost is primarily used for binary classification tasks. It is known for its simplicity and effectiveness, especially with decision stumps (shallow decision trees) as weak learners. It's often used in applications like face detection and object recognition.
Gradient Boosting: Gradient Boosting is more versatile and can be applied to both classification and regression tasks, including complex problems with non-linear relationships. It's widely used in various domains, such as fraud detection, customer churn prediction, and ranking problems, due to its high predictive power. Popular implementations include XGBoost, LightGBM, and CatBoost.


Question 6:Why does CatBoost perform well on categorical features without requiring
extensive preprocessing? Briefly explain its handling of categorical variables.

--> CatBoost performs well on categorical features without extensive preprocessing primarily because of its Ordered Target Encoding method, which addresses the target leakage problem inherent in standard target-based encodings. This approach encodes categorical features into numerical values in a way that prevents the model from "cheating" and seeing the target variable's effect on the very data point it is currently encoding.
CatBoost's handling of categorical variables
CatBoost's approach to handling categorical features is a systematic process of converting categories to numerical values during training. Its core techniques include:
1. Ordered Target Encoding
This is CatBoost's most significant innovation for categorical features. Standard target encoding can lead to data leakage by using the average target value of a category to encode all instances of that category. This allows the model to glimpse the target it is trying to predict and leads to overfitting.
To solve this, CatBoost employs an ordered encoding scheme:
It begins by creating a random permutation (ordering) of the training data.
For each data point, it calculates a numerical statistic for its categorical features using only the data points that came before it in the random permutation.
For example, to encode the City for the 10th row, it would use the average target value of all matching City entries in the first nine rows. For the 11th row, it would use data from the first ten rows, and so on.
To reduce variance from this ordering, CatBoost creates multiple random permutations during training and aggregates the results.
2. Feature combinations
CatBoost can automatically construct new, combined categorical features. For instance, if you have two categorical features like Country and City, CatBoost might create a combined feature (Country, City). This is done in a greedy fashion during tree construction and helps the model capture more complex, high-order dependencies in the data.
3. One-Hot Encoding for low cardinality
For categorical features with a small number of unique values, CatBoost uses a standard one-hot encoding approach. It applies a threshold to decide which categorical features to one-hot encode. This is faster and simpler for features with few categories, such as gender.
By automating these processes internally, CatBoost saves significant time and effort on manual feature engineering and reduces the risk of errors and target leakage, resulting in a more robust and accurate model.


Question 7: KNN Classifier Assignment: Wine Dataset Analysis with
Optimization
Task:
1. Load the Wine dataset (sklearn.datasets.load_wine()).
2. Split data into 70% train and 30% test.
3. Train a KNN classifier (default K=5) without scaling and evaluate using:
a. Accuracy
b. Precision, Recall, F1-Score (print classification report)
4. Apply StandardScaler, retrain KNN, and compare metrics.
5. Use GridSearchCV to find the best K (test K=1 to 20) and distance metric
(Euclidean, Manhattan).

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report

# 1. Load the Wine dataset
wine = load_wine()
X, y = wine.data, wine.target

# 2. Split data into 70% train and 30% test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Train a KNN classifier (default K=5) without scaling and evaluate
print("--- KNN without Scaling ---")
knn_unscaled = KNeighborsClassifier(n_neighbors=5)
knn_unscaled.fit(X_train, y_train)
y_pred_unscaled = knn_unscaled.predict(X_test)

accuracy_unscaled = accuracy_score(y_test, y_pred_unscaled)
report_unscaled = classification_report(y_test, y_pred_unscaled)

print("Accuracy:", accuracy_unscaled)
print("Classification Report:\n", report_unscaled)

# 4. Apply StandardScaler, retrain KNN, and compare metrics
print("\n--- KNN with Scaling ---")
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

knn_scaled = KNeighborsClassifier(n_neighbors=5)
knn_scaled.fit(X_train_scaled, y_train)
y_pred_scaled = knn_scaled.predict(X_test_scaled)

accuracy_scaled = accuracy_score(y_test, y_pred_scaled)
report_scaled = classification_report(y_test, y_pred_scaled)

print("Accuracy:", accuracy_scaled)
print("Classification Report:\n", report_scaled)

# 5. Use GridSearchCV to find the best K (test K=1 to 20) and distance metric
print("\n--- Hyperparameter Tuning with GridSearchCV ---")
param_grid = {'n_neighbors': range(1, 21),
              'metric': ['euclidean', 'manhattan']}

knn_grid = KNeighborsClassifier()

grid_search = GridSearchCV(knn_grid, param_grid, cv=5)
grid_search.fit(X_train_scaled, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best cross-validation score:", grid_search.best_score_)

# Evaluate KNN with best hyperparameters
print("\n--- KNN with Best Parameters ---")
best_k = grid_search.best_params_['n_neighbors']
best_metric = grid_search.best_params_['metric']

knn_best = KNeighborsClassifier(n_neighbors=best_k, metric=best_metric)
knn_best.fit(X_train_scaled, y_train)
y_pred_best = knn_best.predict(X_test_scaled)

accuracy_best = accuracy_score(y_test, y_pred_best)
report_best = classification_report(y_test, y_pred_best)

print("Accuracy:", accuracy_best)
print("Classification Report:\n", report_best)
```

Question 8 : PCA + KNN with Variance Analysis and Visualization
Task:
1. Load the Breast Cancer dataset (sklearn.datasets.load_breast_cancer()).
2. Apply PCA and plot the scree plot (explained variance ratio).
3. Retain 95% variance and transform the dataset.
4. Train KNN on the original data and PCA-transformed data, then compare
accuracy.
5. Visualize the first two principal components using a scatter plot (color by class).

```python
from sklearn.datasets import load_breast_cancer

# 1. Load the Breast Cancer dataset
breast_cancer = load_breast_cancer()
X, y = breast_cancer.data, breast_cancer.target
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Apply PCA
pca = PCA()
pca.fit(X)

# Plot the scree plot (explained variance ratio)
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(pca.explained_variance_ratio_) + 1), pca.explained_variance_ratio_.cumsum(), marker='o', linestyle='--')
plt.title('Scree Plot (Explained Variance Ratio)')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance Ratio')
plt.grid(True)
plt.show()
```

```python
# 3. Retain 95% variance and transform the dataset
# Find the number of components to retain 95% variance
cumulative_variance_ratio = pca.explained_variance_ratio_.cumsum()
n_components_95 = (cumulative_variance_ratio < 0.95).sum() + 1
print(f"Number of components to retain 95% variance: {n_components_95}")

pca_95 = PCA(n_components=n_components_95)
X_pca = pca_95.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Split data for original and PCA-transformed data
X_train_orig, X_test_orig, y_train_orig, y_test_orig = train_test_split(X, y, test_size=0.3, random_state=42)
X_train_pca, X_test_pca, y_train_pca, y_test_pca = train_test_split(X_pca, y, test_size=0.3, random_state=42)

# 4. Train KNN on the original data and PCA-transformed data, then compare accuracy
print("--- KNN on Original Data ---")
knn_original = KNeighborsClassifier()
knn_original.fit(X_train_orig, y_train_orig)
y_pred_original = knn_original.predict(X_test_orig)
accuracy_original = accuracy_score(y_test_orig, y_pred_original)
print("Accuracy on original data:", accuracy_original)

print("\n--- KNN on PCA-Transformed Data ---")
knn_pca = KNeighborsClassifier()
knn_pca.fit(X_train_pca, y_train_pca)
y_pred_pca = knn_pca.predict(X_test_pca)
accuracy_pca = accuracy_score(y_test_pca, y_pred_pca)
print("Accuracy on PCA-transformed data:", accuracy_pca)

print("\n--- Comparison ---")
print(f"Accuracy on original data: {accuracy_original:.4f}")
print(f"Accuracy on PCA-transformed data: {accuracy_pca:.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 5. Visualize the first principal component using a scatter plot (color by class)
plt.figure(figsize=(10, 7))
# Check if X_pca has at least two components
if X_pca.shape[1] >= 2:
    scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis', alpha=0.6)
    plt.ylabel('Principal Component 2')
else:
    # If only one component, plot against a zero array
    scatter = plt.scatter(X_pca[:, 0], np.zeros_like(X_pca[:, 0]), c=y, cmap='viridis', alpha=0.6)
    plt.ylabel('Zero') # Label the y-axis as zero

plt.title('First Principal Component of Breast Cancer Dataset')
plt.xlabel('Principal Component 1')
plt.colorbar(scatter, label='Class')
plt.grid(True)
plt.show()
```

Question 10: KNN with KD-Tree/Ball Tree, Imputation, and Real-World
Data
Task:

Dataset: Pima Indians Diabetes

```python
from sklearn.impute import KNNImputer
import numpy as np
import pandas as pd

# 1. Load the Pima Indians Diabetes dataset (copied from previous cell for independence)
# The dataset is not directly available in sklearn datasets,
# so we'll load it from a common online source.
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.csv"
names = ['pregnancies', 'glucose', 'blood_pressure', 'skin_thickness', 'insulin', 'bmi', 'diabetes_pedigree_function', 'age', 'outcome']
dataset = pd.read_csv(url, names=names)

# The dataset uses 0 to represent missing values for some features.
# We will replace these 0s with NaN for proper imputation.
cols_to_impute = ['glucose', 'blood_pressure', 'skin_thickness', 'insulin', 'bmi']
for col in cols_to_impute:
    dataset[col] = dataset[col].astype(str).replace('0', np.nan).astype(float)


X = dataset.drop('outcome', axis=1)
y = dataset['outcome']

print("Pima Indians Diabetes dataset loaded.")
print("Dataset shape:", dataset.shape)
print("Missing values before imputation:\n", dataset.isnull().sum())

# 2. Use KNN Imputation to fill missing values
imputer = KNNImputer(n_neighbors=5)

# Replace pd.NA with np.nan explicitly before converting to numpy array
X_cleaned = X.replace(pd.NA, np.nan)

# Convert to numpy array
X_imputed_np = imputer.fit_transform(X_cleaned.to_numpy())

# Convert back to DataFrame for easier handling if needed later
X_imputed = pd.DataFrame(X_imputed_np, columns=X.columns)

print("\nMissing values after imputation:\n", pd.DataFrame(X_imputed, columns=X.columns).isnull().sum())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import time

# Split the imputed data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X_imputed, y, test_size=0.3, random_state=42)

# 3. Train KNN using different algorithms and compare training time and accuracy
algorithms = ['brute', 'kd_tree', 'ball_tree']
results = {}

for algo in algorithms:
    print(f"\n--- Training KNN with algorithm: {algo} ---")
    knn = KNeighborsClassifier(algorithm=algo)

    start_time = time.time()
    knn.fit(X_train, y_train)
    end_time = time.time()
    training_time = end_time - start_time
    print(f"Training time: {training_time:.4f} seconds")

    y_pred = knn.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {accuracy:.4f}")

    results[algo] = {'training_time': training_time, 'accuracy': accuracy}

# 4. Compare their training time and accuracy
print("\n--- Comparison of KNN Algorithms ---")
for algo, metrics in results.items():
    print(f"Algorithm: {algo}, Training Time: {metrics['training_time']:.4f}s, Accuracy: {metrics['accuracy']:.4f}")
```

```python
from sklearn.ensemble import RandomForestClassifier
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# Identify the two most important features
# Using a RandomForestClassifier to get feature importances as a proxy
model = RandomForestClassifier(random_state=42)
model.fit(X_imputed, y)

feature_importances = pd.Series(model.feature_importances_, index=X_imputed.columns)
two_most_important_features = feature_importances.nlargest(2).index.tolist()

print("Two most important features:", two_most_important_features)

# Select the best-performing algorithm based on accuracy (assuming higher is better)
best_algo = max(results, key=lambda algo: results[algo]['accuracy'])
print(f"Best performing algorithm based on accuracy: {best_algo}")

# Train the best-performing KNN model on the imputed data with the two most important features
X_imputed_two_features = X_imputed[two_most_important_features]
X_train_two_features, X_test_two_features, y_train_two_features, y_test_two_features = train_test_split(
    X_imputed_two_features, y, test_size=0.3, random_state=42
)

knn_best_two_features = KNeighborsClassifier(algorithm=best_algo)
knn_best_two_features.fit(X_train_two_features, y_train_two_features)

# Plot the decision boundary
plt.figure(figsize=(10, 7))

# Create a meshgrid of points to plot the decision boundary
h = .02  # step size in the mesh
x_min, x_max = X_imputed_two_features.iloc[:, 0].min() - 1, X_imputed_two_features.iloc[:, 0].max() + 1
y_min, y_max = X_imputed_two_features.iloc[:, 1].min() - 1, X_imputed_two_features.iloc[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                     np.arange(y_max, y_min, -h)) # Corrected to plot from y_max to y_min

# Predict the class for each point in the meshgrid
# Convert meshgrid points to DataFrame with correct column names
meshgrid_points = pd.DataFrame(np.c_[xx.ravel(), yy.ravel()], columns=two_most_important_features)
Z = knn_best_two_features.predict(meshgrid_points)
Z = Z.reshape(xx.shape)

# Plot the decision boundary
plt.contourf(xx, yy, Z, cmap=plt.cm.RdYlBu, alpha=0.8)

# Plot the training points
scatter = plt.scatter(X_train_two_features.iloc[:, 0], X_train_two_features.iloc[:, 1], c=y_train_two_features, cmap=plt.cm.RdYlBu, edgecolor='k', s=20)
plt.title('KNN Decision Boundary with Two Most Important Features')
plt.xlabel(two_most_important_features[0])
plt.ylabel(two_most_important_features[1])
plt.colorbar(scatter, label='Outcome')
plt.show()
```
